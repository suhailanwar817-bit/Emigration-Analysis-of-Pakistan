### Importing Raw Data

```python
import pandas as pd

df = pd.read_csv(r"Raw_Data.csv", encoding="latin-1")
df.head()
```

### Data Dimensions

```python
print("Number of Rows: ", df.shape[0])
print("Number of Columns: ", df.shape[1])
```

### Total vs Unique Districts

```python
print("Total Districts: ", df.shape[1])
print("Unique Districts: ", df['Districts'].nunique())
```

### Finding Column names

```python
df.columns.tolist()
```

### Identify district names containing extra spaces

```python
whitespace = df[df['Districts'] != df['Districts'].str.strip()]

display(whitespace[['Districts']])

print("Districts with extra spaces:", len(whitespace))
```

### Removing Trailing and Leading Spaces

```python
before = df['Districts'].str.len() - df['Districts'].str.strip().str.len()

print("Extra Spaces Before: ", before.sum())
      
df['Districts'] = df['Districts'].str.strip()

After = df['Districts'].str.len() - df['Districts'].str.strip().str.len()

print("Extra Spaces After: ", After.sum())
```

### Finding Missing Values

```python
missing = pd.DataFrame({
    'Missing Values': df.isnull().sum(),
    'Missing Percentage': (df.isnull().mean() * 100).round(2)
})

display(missing)
```

### Finding Duplicates

```python
duplicates = df.duplicated().sum()

print("Total duplicates:", duplicates)
```
